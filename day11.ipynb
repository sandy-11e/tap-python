```python
from pathlib import Path

INPUT_FOLDER = Path(".")
```

```python
# 7. Loop over all Olist CSV files and print a summary for each.

import pandas as pd

csv_files = list(INPUT_FOLDER.glob("olist*.csv"))

for file in csv_files:
    df = pd.read_csv(file)

    print("\nFile name:", file.name)
    print("Rows and columns:", df.shape)
    print("Total null values:", df.isnull().sum().sum())

```

```python
# 10. Make a copy of one file with ';' as the delimiter and handle both formats.

import pandas as pd

input_file = "olist_orders_dataset.csv"
output_file = "olist_order_items_dataset.csv"
try:
    df = pd.read_csv(input_file, sep=",")
except:
    df = pd.read_csv(input_file, sep="\t")

df.to_csv(output_file, sep=";", index=False)

print("File copied successfully!")
```
