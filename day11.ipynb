```python
# 7. Loop over all Olist CSV files and print a summary for each.

import pandas as pd
import glob

lst=glob.glob("olist*.csv")+["product_category_name_translation.csv"]

for file in lst:
    df=pd.read_csv(file)
    print("\nfile name : ",file)
    print("no.of Rows and columns : ",df.shape)
    print("null count :\n",df.isnull().sum().sum())


```
