```python
NULLABLE_FIELDS = {
    "olist_orders_dataset.csv": [
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date"
    ],

    "olist_order_reviews_dataset.csv": [
        "review_comment_title",
        "review_comment_message",
        "review_answer_timestamp"
    ],

    "olist_products_dataset.csv": [
        "product_category_name"
    ]
}
```

```python
def validate_rows(df, filename):

    errors = []

    # Fields that must contain values
    nullable = NULLABLE_FIELDS.get(filename, [])

    mandatory_fields = [
        col for col in REQUIRED_FIELDS[filename]
        if col not in nullable
    ]

    for index, row in df.iterrows():

        row_errors = []

        for column in mandatory_fields:

            value = row[column]

            if pd.isna(value) or str(value).strip() == "":
                row_errors.append(f"{column} is missing")

        if row_errors:

            record = row.to_dict()

            record["source_file"] = filename
            record["row_number"] = index + 2
            record["reason"] = "; ".join(row_errors)

            errors.append(record)

    return errors
```

```python
all_errors = []

for filename in REQUIRED_FIELDS:

    file_path = DATA_FOLDER / filename

    df = pd.read_csv(file_path)

    errors = validate_rows(df, filename)

    all_errors.extend(errors)

    print(
        f"{filename}: "
        f"{len(df)} rows checked, "
        f"{len(errors)} errors found"
    )
```

```python
def validate_rows(df, filename):

    errors = []

    nullable = NULLABLE_FIELDS.get(filename, [])

    mandatory_fields = [
        col for col in REQUIRED_FIELDS[filename]
        if col not in nullable
    ]

    for index, row in df.iterrows():

        row_errors = []

        # 1. Mandatory field validation
        for column in mandatory_fields:

            value = row[column]

            if pd.isna(value) or str(value).strip() == "":
                row_errors.append(f"{column} is missing")

        # 2. Numeric validation
        for column in NUMERIC_COLUMNS:

            if column in df.columns and pd.notna(row[column]):

                try:
                    float(row[column])

                except (ValueError, TypeError):
                    row_errors.append(
                        f"{column} must be numeric"
                    )

        # 3. Date validation
        for column in DATE_COLUMNS:

            if column in df.columns and pd.notna(row[column]):

                if str(row[column]).strip() != "":

                    parsed_date = pd.to_datetime(
                        row[column],
                        errors="coerce"
                    )

                    if pd.isna(parsed_date):
                        row_errors.append(
                            f"{column} contains invalid date"
                        )

        # Collect rejected records
        if row_errors:

            record = row.to_dict()
            record["source_file"] = filename
            record["row_number"] = index + 2
            record["reason"] = "; ".join(row_errors)

            errors.append(record)

    return errors
```

```python
class SchemaError(Exception):
    """Raised when the CSV schema is invalid."""
    pass


class FileFormatError(Exception):
    """Raised when the file is missing, empty or unreadable."""
    pass
```

```python
try:
    raise SchemaError("Required column 'order_id' is missing")

except SchemaError as e:
    print("Schema Error:", e)


try:
    raise FileFormatError("CSV file is empty")

except FileFormatError as e:
    print("File Format Error:", e)
```

```python
def read_csv_safely(file_path):

    file_path = Path(file_path)

    # Check file existence
    if not file_path.exists():
        raise FileFormatError(
            f"File not found: {file_path}"
        )

    # Check empty file
    if file_path.stat().st_size == 0:
        raise FileFormatError(
            f"File is empty: {file_path}"
        )

    try:
        df = pd.read_csv(file_path)

    except pd.errors.EmptyDataError:
        raise FileFormatError(
            f"No data found in: {file_path}"
        )

    except (pd.errors.ParserError, UnicodeDecodeError) as e:
        raise FileFormatError(
            f"Invalid CSV format: {e}"
        )

    # Check empty DataFrame
    if df.empty:
        raise FileFormatError(
            f"No records found: {file_path}"
        )

    return df
```

```python
def validate_schema(df, filename):

    if filename not in REQUIRED_FIELDS:
        raise SchemaError(
            f"No schema defined for {filename}"
        )

    missing_columns = set(REQUIRED_FIELDS[filename]) - set(df.columns)

    if missing_columns:
        raise SchemaError(
            f"Missing columns in {filename}: {missing_columns}"
        )

    return True
```

```python
MAX_REJECTION_PERCENTAGE = 20
```

```python
all_errors = []

for filename in REQUIRED_FIELDS:

    file_path = DATA_FOLDER / filename

    try:
        df = read_csv_safely(file_path)

        validate_schema(df, filename)

        errors = validate_rows(df, filename)

        all_errors.extend(errors)

        print(
            f"{filename}: "
            f"{len(df)} rows checked | "
            f"{len(errors)} rejected"
        )

    except (SchemaError, FileFormatError) as e:

        print(f"\nERROR: {e}")
        continue
```

```python
file_summaries = []
all_errors = []

for filename in REQUIRED_FIELDS:

    file_path = DATA_FOLDER / filename

    try:
        df = read_csv_safely(file_path)

        validate_schema(df, filename)

        errors = validate_rows(df, filename)

        all_errors.extend(errors)

        total_rows = len(df)
        rejected_rows = len(errors)
        valid_rows = total_rows - rejected_rows

        rejection_percentage = (
            rejected_rows / total_rows
        ) * 100

        # Check rejection threshold
        if rejection_percentage > MAX_REJECTION_PERCENTAGE:
            status = "FAILED"
        else:
            status = "PASSED"

        file_summaries.append({
            "file": filename,
            "total_rows": total_rows,
            "valid_rows": valid_rows,
            "rejected_rows": rejected_rows,
            "rejection_percentage": round(
                rejection_percentage, 2
            ),
            "status": status
        })

        print(f"\nFile: {filename}")
        print(f"Total: {total_rows}")
        print(f"Valid: {valid_rows}")
        print(f"Rejected: {rejected_rows}")
        print(f"Rejection: {rejection_percentage:.2f}%")
        print(f"Status: {status}")

        if status == "FAILED":
            print("File rejected due to excessive bad records.")

    except (SchemaError, FileFormatError) as e:

        print(f"ERROR: {e}")

        file_summaries.append({
            "file": filename,
            "total_rows": 0,
            "valid_rows": 0,
            "rejected_rows": 0,
            "rejection_percentage": 0,
            "status": "ERROR"
        })
```

```python
summary_df = pd.DataFrame(file_summaries)

display(summary_df)

print("\n========== FINAL VALIDATION REPORT ==========")

print("Total files:", len(summary_df))
print("Passed files:", (summary_df["status"] == "PASSED").sum())
print("Failed files:", (summary_df["status"] == "FAILED").sum())
print("Error files:", (summary_df["status"] == "ERROR").sum())

print("Total rows:", summary_df["total_rows"].sum())
print("Valid rows:", summary_df["valid_rows"].sum())
print("Rejected rows:", summary_df["rejected_rows"].sum())
```
